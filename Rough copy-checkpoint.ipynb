```python
class Person:
    def __init__(self,gender,age,name,height, weight,bloodgroup):
        # public (can be accessed anywhere)
        self.gender = gender
        self.age = age
        self.name = name
        self.height = height
        self.weight = weight
        self._bloodgroup = bloodgroup
    def eat(self):
        print(f"{self.name} a {self.gender} with height {self.height} and with  weight {self.weight} is eating")
    def sleep(self):
        print(f"{self.name} a {self.gender} with height {self.height} and with weight {self.weight} is sleeping")
    def walk (self):
        print(f"{self.name} a {self.gender} with height {self.height} and with weight {self.weight} is walking ")
    def get_blood_group(self):
        return self._bloodgroup
    def set_blood_group(self,new_blood_group):        
        blood_groups = ['A+','A-','B+','B-','O+','O-','AB+']
        if new_blood_group in blood_groups:
            self._blood_group = new_blood_group
        else:
            return "Invalid Blood group"
```
