```python
import bcrypt

def hash_password(password):
    salt = bcrypt.gensalt()
    return bcrypt.hashpw(password.encode(), salt)

def verify_password(password, hashed):
    return bcrypt.checkpw(password.encode(), hashed)
```

```python
import streamlit as st
import sqlite3
import pandas as pd

conn = sqlite3.connect("database/nearlearn.db")

st.set_page_config(layout="wide")

st.title("🔍 Find Tutors")

col1,col2,col3,col4 = st.columns(4)

subject = col1.text_input("Subject")

city = col2.text_input("City")

max_fee = col3.number_input("Maximum Fee",0,100000,5000)

rating = col4.slider("Minimum Rating",0.0,5.0,4.0)

query = """
SELECT
name,
subject,
experience,
fee,
city,
rating,
verified
FROM tutors
WHERE fee<=?
AND rating>=?
"""

df = pd.read_sql_query(
    query,
    conn,
    params=(max_fee,rating)
)

if subject:
    df = df[df["subject"].str.contains(subject,case=False)]

if city:
    df = df[df["city"].str.contains(city,case=False)]

st.write(f"Found {len(df)} tutors")

for _,row in df.iterrows():

    with st.container():

        st.markdown("---")

        c1,c2 = st.columns([1,4])

        with c1:
            st.image("assets/default_teacher.png",width=120)

        with c2:

            st.subheader(row["name"])

            st.write(f"📘 Subject : {row['subject']}")

            st.write(f"🏙 City : {row['city']}")

            st.write(f"⭐ Rating : {row['rating']}")

            st.write(f"💰 Fee : PKR {row['fee']}")

            st.write(f"👨‍🏫 Experience : {row['experience']} years")

            if row["verified"]:

                st.success("Verified Tutor")

            c3,c4 = st.columns(2)

            with c3:

                st.button("View Profile",key=row["name"])

            with c4:

                st.button("Book Now",key=row["email"])
```

```python
import hashlib
import hmac
import time
import requests

# JazzCash Sandbox Credentials
MERCHANT_ID = "YOUR_MERCHANT_ID"
PASSWORD = "YOUR_PASSWORD"
INTEGRITY_SALT = "YOUR_INTEGRITY_SALT"

RETURN_URL = "http://localhost:8501"
CURRENCY = "PKR"


def create_payment(amount):

    txn_ref = "NL" + str(int(time.time()))

    payload = {

        "pp_Version": "1.1",

        "pp_TxnType": "MWALLET",

        "pp_Language": "EN",

        "pp_MerchantID": MERCHANT_ID,

        "pp_Password": PASSWORD,

        "pp_TxnRefNo": txn_ref,

        "pp_Amount": str(amount * 100),

        "pp_Currency": CURRENCY,

        "pp_ReturnURL": RETURN_URL

    }

    message = "&".join(str(v) for v in payload.values())

    secure_hash = hmac.new(
        INTEGRITY_SALT.encode(),
        message.encode(),
        hashlib.sha256
    ).hexdigest()

    payload["pp_SecureHash"] = secure_hash

    return payload
```

```python
import sqlite3
import pandas as pd

conn = sqlite3.connect(
    "database/nearlearn.db",
    check_same_thread=False
)

def recommend(subject, city, budget):

    tutors = pd.read_sql_query(
        """
        SELECT *
        FROM tutors
        WHERE
        subject=?
        AND city=?
        AND fee<=?
        AND available=1
        """,
        conn,
        params=(subject, city, budget)
    )

    if tutors.empty:
        return tutors

    tutors["score"] = (
        tutors["rating"] * 40
        + tutors["verified"] * 20
        + tutors["completed_classes"] * 0.5
        + tutors["experience"] * 3
        + tutors["total_reviews"] * 0.2
        - tutors["fee"] / 1000
        - tutors["response_time"] * 0.1
    )

    tutors = tutors.sort_values(
        by="score",
        ascending=False
    )

    return tutors
```

```python
students = conn.execute(
    "SELECT COUNT(*) FROM students"
).fetchone()[0]

tutors = conn.execute(
    "SELECT COUNT(*) FROM tutors"
).fetchone()[0]

centers = conn.execute(
    "SELECT COUNT(*) FROM education_centers"
).fetchone()[0]

bookings = conn.execute(
    "SELECT COUNT(*) FROM bookings"
).fetchone()[0]

c1,c2,c3,c4 = st.columns(4)

c1.metric("👨‍🎓 Students", students)

c2.metric("👨‍🏫 Tutors", tutors)

c3.metric("🏫 Centers", centers)

c4.metric("📚 Bookings", bookings)
```

```python
payments = pd.read_sql_query("""

SELECT *

FROM payments

""",conn)

total = payments["amount"].sum()

commission = payments["commission"].sum()

provider = payments["provider_amount"].sum()

a,b,c = st.columns(3)

a.metric("💰 Total Revenue",f"PKR {total:,}")

b.metric("🏢 NearLearn Commission",f"PKR {commission:,}")

c.metric("👨‍🏫 Tutor Earnings",f"PKR {provider:,}")
```

```python
payments["payment_date"] = pd.to_datetime(
    payments["payment_date"]
)

payments["Month"] = payments[
    "payment_date"
].dt.strftime("%Y-%m")

monthly = payments.groupby(
    "Month"
)["amount"].sum().reset_index()

fig = px.line(

monthly,

x="Month",

y="amount",

title="Monthly Revenue"

)

st.plotly_chart(
    fig,
    use_container_width=True
)
```

```python
booking = pd.read_sql_query("""

SELECT

status,

COUNT(*) total

FROM bookings

GROUP BY status

""",conn)

fig = px.pie(

booking,

names="status",

values="total",

title="Booking Status"

)

st.plotly_chart(fig)
```

```python
subjects = pd.read_sql_query("""

SELECT

subject,

COUNT(*) total

FROM bookings

GROUP BY subject

ORDER BY total DESC

LIMIT 10

""",conn)

fig = px.bar(

subjects,

x="subject",

y="total",

title="Popular Subjects"

)

st.plotly_chart(fig)
```

```python
top = pd.read_sql_query("""

SELECT

provider_email,

COUNT(*) classes

FROM bookings

GROUP BY provider_email

ORDER BY classes DESC

LIMIT 10

""",conn)

st.subheader("🏆 Top Tutors")

st.dataframe(top)
```
